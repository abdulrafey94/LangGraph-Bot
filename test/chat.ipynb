```python
import requests
import random
from IPython.display import display, Markdown

BASE_URL = "http://127.0.0.1:8000"

thread_id = f"thread-{str(random.randint(10000, 99999))}"
print(f"Initializing Chat Thread: {thread_id}")

print("Please enter your travel query (type exit to close):\n")
while True:
    message = input("Query:")
    payload = {"thread_id": thread_id, "message": message}

    if message.lower() == "exit":
       break
    response = requests.post(f"{BASE_URL}/chat", json=payload)
    assert response.status_code == 200, f"Error {response.status_code}: {response.text}"
    
    data = response.json()
    display(Markdown(f"\nResponse: {data['reply']}\n"))
```
